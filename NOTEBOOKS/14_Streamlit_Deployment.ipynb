# Insider Threat Detection Web Application

This notebook prepares the Streamlit web application for deploying the trained Random Forest model.

The web application allows users to:

- Predict insider threats for a single employee
- Upload CSV files for batch prediction
- Display prediction probabilities
- Visualize risk levels
- Download prediction results

The application is built using Streamlit.

```python
from pathlib import Path
import os
```

# Create Project Folders

```python
STREAMLIT_PATH = Path("../STREAMLIT")

STREAMLIT_PATH.mkdir(
    parents=True,
    exist_ok=True
)

(STREAMLIT_PATH / "assets").mkdir(
    exist_ok=True
)

print("Folder Created Successfully")
```

# Verify Folder Structure

```python
print("STREAMLIT Folder Exists :", STREAMLIT_PATH.exists())

print()

print("Folder Contents")

for file in STREAMLIT_PATH.iterdir():
    print(file.name)
```

# Create Streamlit Files

```python
files = [

    "app.py",

    "predict.py",

    "utils.py",

    "style.css",

    "requirements.txt"

]

for file in files:

    (STREAMLIT_PATH / file).touch(
        exist_ok=True
    )

print("Files Created Successfully")
```

# Verify Files

```python
for file in STREAMLIT_PATH.iterdir():

    print(file.name)
```

# Verify Saved Models

```python
MODEL_PATH = Path("../MODELS")

print("MODEL Folder Exists :", MODEL_PATH.exists())

print()

print("Models")

for file in MODEL_PATH.iterdir():

    print(file.name)
```

# Load Random Forest Model

```python
import joblib

random_forest = joblib.load(

    MODEL_PATH /

    "random_forest_model.pkl"

)

print(random_forest)
```

# Create app.py

The Streamlit application is maintained **directly in the files**, not generated by this notebook:

- `STREAMLIT/app.py` - UI (Home, Single Prediction, Batch Prediction, About)
- `STREAMLIT/predict.py` - prediction logic
- `STREAMLIT/utils.py` - model / feature loading and CSS
- `STREAMLIT/style.css` - custom styling

This notebook does **NOT** rewrite those files, so re-running it is safe and never clobbers the hand-edited application code. This section only verifies that the application files exist.


```python
app_files = [

    "app.py",

    "predict.py",

    "utils.py",

    "style.css",

    "requirements.txt"

]

for file in app_files:

    path = STREAMLIT_PATH / file

    print(
        file,
        "Exists :",
        path.exists()
    )
```

# Create requirements.txt

```python
requirements = """
streamlit
pandas
numpy
matplotlib
scikit-learn
joblib
shap
"""

with open(
    STREAMLIT_PATH / "requirements.txt",
    "w"
) as file:

    file.write(requirements)

print("requirements.txt Created Successfully")
```

# Create Custom CSS

```python
css = """
.main {

    background-color:#f5f5f5;

}

.stButton>button{

    width:100%;

    border-radius:10px;

    height:45px;

}

h1{

    color:#0E76A8;

}

"""

with open(

    STREAMLIT_PATH/"style.css",

    "w"

) as file:

    file.write(css)

print("CSS Created Successfully")
```

```python
print("Open Terminal")

print()

print("cd STREAMLIT")

print()

print("streamlit run app.py")
```
