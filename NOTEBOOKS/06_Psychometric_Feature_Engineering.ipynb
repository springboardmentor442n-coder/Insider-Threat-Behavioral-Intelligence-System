# Rename User ID Column

The psychometric dataset uses the column name **user_id**, whereas all other feature datasets use **user**.

To ensure consistency across all datasets, we rename **user_id** to **user**.

```python
psychometric = psychometric.rename(
    columns={"user_id": "user"}
)
```

```python
psychometric.columns
```

```python
# Check Personality Score Statistics
psychometric.describe()
```

```python
# Select Final Personality Features
psychometric_features = psychometric[
    [
        "user",
        "O",
        "C",
        "E",
        "A",
        "N"
    ]
]
```

```python
# Check for Missing Values
psychometric_features.isnull().sum()
```

```python
# Replace Missing Values (If Any)
psychometric_features = psychometric_features.fillna(0)
```

```python
psychometric_features.isnull().sum()
```

```python
# Save Psychometric Features
FEATURE_PATH = Path("../DATA/features")
FEATURE_PATH.mkdir(parents=True, exist_ok=True)

psychometric_features.to_csv(
    FEATURE_PATH / "psychometric_features.csv",
    index=False
)
```
