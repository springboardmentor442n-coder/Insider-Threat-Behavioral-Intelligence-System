```python
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
DATA_PATH = Path("../DATA/raw/CERT_R6.2/data")
```

```python
file = pd.read_csv(DATA_PATH / "file.csv")
```

```python
file.info()
```

```python
#check missing values
file.isnull().sum()
```

```python
file.duplicated().sum()
```

```python
# Convert Date Column to Datetime
file["date"] = pd.to_datetime(file["date"])
```

```python
file["date"].head()
```

```python
file["date"].dtype
```

```python
# Extract Date and Time Features
file["year"] = file["date"].dt.year
file["month"] = file["date"].dt.month
file["day"] = file["date"].dt.day
file["hour"] = file["date"].dt.hour
file["weekday"] = file["date"].dt.day_name()
```

```python
file[
    ["date", "year", "month", "day", "hour", "weekday"]
].head()
```

```python
# Create After-Hours and Weekend Features
file["after_hours"] = (
    (file["hour"] < 8) |
    (file["hour"] >= 18)
)

file["weekend_activity"] = file["weekday"].isin(
    ["Saturday", "Sunday"]
)
```

```python
file[
    ["hour", "weekday", "after_hours", "weekend_activity"]
].head()
```

```python
# Explore File Activities
file["activity"].value_counts()
```

```python
# Create Initial Employee File Profile
file_profile = (
    file.groupby("user")
        .size()
        .reset_index(name="total_file_events")
)
```

```python
file_profile.head()
```

```python
# Calculate File Open Count
file_open = (
    file[file["activity"] == "File Open"]
    .groupby("user")
    .size()
    .reset_index(name="file_open_count")
)
```

```python
file_open.head()
```

```python
# Merge File Open Count
file_profile = file_profile.merge(
    file_open,
    on="user",
    how="left"
)
file_profile.head()
```

```python
# Calculate File Copy Count
file_copy = (
    file[file["activity"] == "File Copy"]
    .groupby("user")
    .size()
    .reset_index(name="file_copy_count")
)
```

```python
file_copy.head()
```

```python
# Merge File Copy Count
file_profile = file_profile.merge(
    file_copy,
    on="user",
    how="left"
)
```

```python
file_profile.head()
```

```python
# Calculate File Write Count
file_write = (
    file[file["activity"] == "File Write"]
    .groupby("user")
    .size()
    .reset_index(name="file_write_count")
)
```

```python
# Merge File Write Count
file_profile = file_profile.merge(
    file_write,
    on="user",
    how="left"
)
```

```python
file_profile.head()
```

```python
# Calculate File Delete Count
file_delete = (
    file[file["activity"] == "File Delete"]
    .groupby("user")
    .size()
    .reset_index(name="file_delete_count")
)
```

```python
# Merge File Delete Count
file_profile = file_profile.merge(
    file_delete,
    on="user",
    how="left"
)
```

```python
file_profile.head()
```

```python
# Calculate Files Copied to Removable Media
to_removable = (
    file[file["to_removable_media"]]
    .groupby("user")
    .size()
    .reset_index(name="files_to_removable_media")
)
```

```python
to_removable.head()
```

```python
# Calculate Files Copied From Removable Media
from_removable = (
    file[file["from_removable_media"]]
    .groupby("user")
    .size()
    .reset_index(name="files_from_removable_media")
)
```

```python
from_removable.head()
```

```python
# Merge Removable Media Features
file_profile = file_profile.merge(
    to_removable,
    on="user",
    how="left"
)

file_profile = file_profile.merge(
    from_removable,
    on="user",
    how="left"
)
```

```python
file_profile.head()
```

```python
# Calculate Unique Files Accessed
unique_files = (
    file.groupby("user")["filename"]
        .nunique()
        .reset_index(name="unique_files")
)
```

```python
unique_files.head()
```

```python
# Calculate Unique Computers Used
unique_pcs = (
    file.groupby("user")["pc"]
        .nunique()
        .reset_index(name="unique_pcs")
)
```

```python
unique_pcs.head()
```

```python
# Calculate After-Hours File Activity
after_hours_activity = (
    file[file["after_hours"]]
    .groupby("user")
    .size()
    .reset_index(name="after_hours_file_activity")
)
```

```python
# Calculate Weekend File Activity
weekend_files = (
    file[file["weekend_activity"]]
    .groupby("user")
    .size()
    .reset_index(name="weekend_file_count")
)
```

```python
weekend_files.head()
```

```python
# Merge Unique Files Feature
file_profile = file_profile.merge(
    unique_files,
    on="user",
    how="left"
)
```

```python
file_profile.head()
```

```python
# Merge Unique PCs Feature
file_profile = file_profile.merge(
    unique_pcs,
    on="user",
    how="left"
)
```

```python
file_profile.head()
```

```python
# Merge After-Hours File Activity
file_profile = file_profile.merge(
    after_hours_activity,
    on="user",
    how="left"
)
```

```python
# Merge Weekend File Activity
file_profile = file_profile.merge(
    weekend_files,
    on="user",
    how="left"
)
```

```python
file_profile.head()
```

```python
# Fill Missing Values
file_profile = file_profile.fillna(0)
```

```python
file_profile.isnull().sum()
```

```python
# Verify the Final File Profile
file_profile.head()
```

```python
file_profile.info()
```

```python
# Save File Features
FEATURE_PATH = Path("../DATA/features")
FEATURE_PATH.mkdir(parents=True, exist_ok=True)

file_profile.to_csv(
    FEATURE_PATH / "file_features.csv",
    index=False
)
```
