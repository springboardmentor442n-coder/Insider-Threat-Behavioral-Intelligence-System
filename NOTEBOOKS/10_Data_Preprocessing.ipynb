# Data Preprocessing

Data preprocessing is the process of cleaning and transforming raw data into a format suitable for Machine Learning algorithms.

In this notebook we will:

- Remove unnecessary columns
- Handle missing values
- Separate features and labels
- Scale numerical features
- Handle class imbalance
- Split into training and testing datasets

```python
import pandas as pd
import numpy as np

from pathlib import Path

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
DATA_PATH = Path("../DATA/processed")

df = pd.read_csv(
    DATA_PATH / "labeled_dataset.csv"
)

print("Dataset Shape:", df.shape)
```

```python
print("=" * 50)
print("Missing Values")
print("=" * 50)

print(df.isnull().sum())

print("\n")

print("=" * 50)
print("Duplicate Rows")
print("=" * 50)

print(df.duplicated().sum())
```

```python
# Remove Unnecessary Columns
# Remove the user identifier

df = df.drop(columns=["user"])

print("Updated Shape:", df.shape)

print(df.columns.tolist())
```

# Separate Features and Target Variable

Machine Learning models require:

- Features (X): Input variables used for prediction.
- Target (y): Output variable that the model predicts.

```python
# Separate Features and Target

X = df.drop(columns=["label"])

y = df["label"]

print("Feature Matrix Shape:", X.shape)
print("Target Shape:", y.shape)
```

# Split Dataset into Training and Testing Sets

The dataset is divided into:

- Training Set (80%)
- Testing Set (20%)

The training set is used to train the model, while the testing set evaluates how well the model performs on unseen data.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Features :", X_train.shape)
print("Testing Features  :", X_test.shape)

print("Training Labels   :", y_train.shape)
print("Testing Labels    :", y_test.shape)
```

# Feature Scaling

Standardization transforms numerical features so they have:

- Mean = 0
- Standard Deviation = 1

Many Machine Learning algorithms perform better when features are on a similar scale.

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)

print("Training Data Shape :", X_train_scaled.shape)
print("Testing Data Shape  :", X_test_scaled.shape)
```

```python
# Verify the Scaled Data
print("First Five Rows of Scaled Training Data:")

print(X_train_scaled[:5])
```

# Handle Class Imbalance using SMOTE

SMOTE (Synthetic Minority Over-sampling Technique) creates synthetic examples of the minority class instead of simply duplicating existing records.

Benefits:

- Reduces class imbalance
- Improves model learning
- Prevents bias toward the majority class

```python
from imblearn.over_sampling import SMOTE
```

# Apply SMOTE on Training Data

SMOTE is applied **only** to the training dataset.

The testing dataset must remain unchanged so that evaluation reflects real-world conditions.

```python
smote = SMOTE(
    random_state=42,
    k_neighbors=2
)

X_train_balanced, y_train_balanced = smote.fit_resample(
    X_train_scaled,
    y_train
)

print("Balanced Training Feature Shape :", X_train_balanced.shape)
print("Balanced Training Label Shape   :", y_train_balanced.shape)
```

# Verify Class Distribution After SMOTE

Compare the class distribution before and after applying SMOTE.

```python
print("Before SMOTE")
print(y_train.value_counts())

print("\nAfter SMOTE")
print(pd.Series(y_train_balanced).value_counts())
```

# Save the Preprocessed Dataset

Save the processed training and testing datasets for model training.

```python
import os

OUTPUT_PATH = "../DATA/processed"

os.makedirs(OUTPUT_PATH, exist_ok=True)

pd.DataFrame(X_train_balanced).to_csv(
    f"{OUTPUT_PATH}/X_train.csv",
    index=False
)

pd.DataFrame(X_test_scaled).to_csv(
    f"{OUTPUT_PATH}/X_test.csv",
    index=False
)

pd.Series(y_train_balanced).to_csv(
    f"{OUTPUT_PATH}/y_train.csv",
    index=False
)

pd.Series(y_test).to_csv(
    f"{OUTPUT_PATH}/y_test.csv",
    index=False
)

print("✅ Processed datasets saved successfully!")
```
