# Merge All Features

This notebook combines all engineered feature datasets into a single employee-level dataset.

Every row represents one employee, while every column represents one behavioral or personality feature.

```python
import pandas as pd
from pathlib import Path
```

```python
FEATURE_PATH = Path("../DATA/features")
```

```python
# Load All Feature Files
employee = pd.read_csv(
    FEATURE_PATH / "employee_profile.csv"
)

device = pd.read_csv(
    FEATURE_PATH / "device_features.csv"
)

email = pd.read_csv(
    FEATURE_PATH / "email_features.csv"
)

file = pd.read_csv(
    FEATURE_PATH / "file_features.csv"
)

http = pd.read_csv(
    FEATURE_PATH / "http_features.csv"
)

psychometric = pd.read_csv(
    FEATURE_PATH / "psychometric_features.csv"
)
```

```python
# Verify Loaded Datasets
print("Employee :", employee.shape)
print("Device :", device.shape)
print("Email :", email.shape)
print("File :", file.shape)
print("HTTP :", http.shape)
print("Psychometric :", psychometric.shape)
```

# Merge Employee and Device Features

The Employee Profile dataset will be used as the base dataset.

We first merge the device features using the common **user** column.

```python
merged_data = employee.merge(
    device,
    on="user",
    how="left"
)
```

```python
# Merge Email Features
merged_data = merged_data.merge(
    email,
    on="user",
    how="left"
)
```

```python
merged_data.head()
```

```python
# Merge File Features
merged_data = merged_data.merge(
    file,
    on="user",
    how="left"
)
```

```python
# Merge HTTP Features
merged_data = merged_data.merge(
    http,
    on="user",
    how="left"
)
merged_data.head()
```

```python
# Verify Intermediate Merged Dataset
merged_data.head()
```

```python
merged_data.info()
```

```python
merged_data.shape
```

# Merge Psychometric Features

The final step is to merge the psychometric (Big Five Personality) features with the behavioral dataset.

This creates a complete employee profile containing both behavioral and personality information.

```python
merged_data = merged_data.merge(
    psychometric,
    on="user",
    how="left"
)
merged_data.head()
```

```python
# Check for Duplicate Employees
duplicate_users = merged_data["user"].duplicated().sum()

print("Duplicate Users:", duplicate_users)
```

```python
# Handle Missing Values
merged_data = merged_data.fillna(0)
```

```python
# Save Final Merged Dataset
FEATURE_PATH = Path("../DATA/features")
FEATURE_PATH.mkdir(parents=True, exist_ok=True)

merged_data.to_csv(
    FEATURE_PATH / "merged_features.csv",
    index=False
)
```
