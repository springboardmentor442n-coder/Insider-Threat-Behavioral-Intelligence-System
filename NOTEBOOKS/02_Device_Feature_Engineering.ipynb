## 02. Device Feature Engineering

### Objective

- Load device.csv
- Understand device usage behavior
- Engineer employee-level USB/device features
- Save device_features.csv

```python
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
# Define Dataset Path
DATA_PATH = Path("../DATA/raw/CERT_R6.2/data")
```

```python
# Load Device Dataset
device = pd.read_csv(DATA_PATH / "device.csv")
```

```python
# Check Activity Types
device["activity"].value_counts()
```

```python
# Convert Date Column to Datetime
device["date"] = pd.to_datetime(device["date"])
```

```python
# Extract Date Features
device["year"] = device["date"].dt.year
device["month"] = device["date"].dt.month
device["day"] = device["date"].dt.day
device["hour"] = device["date"].dt.hour
device["weekday"] = device["date"].dt.day_name()
```

```python
device.head()
```

```python
# Create After Hours Feature
device["after_hours"] = (
    (device["hour"] < 8) |
    (device["hour"] >= 18)
)
```

```python
# Create Weekend Activity Feature
device["weekend_activity"] = device["weekday"].isin(
    ["Saturday", "Sunday"]
)
```

```python
device.head()
```

```python
# Create Employee Device Profile
device_profile = (
    device.groupby("user")
          .size()
          .reset_index(name="total_device_events")
)
```

```python
# Calculate Total Connect Events Per User
total_connects = (
    device[device["activity"] == "Connect"]
    .groupby("user")
    .size()
    .reset_index(name="total_connects")
)
```

```python
# Merge Total Connect Events
device_profile = device_profile.merge(
    total_connects,
    on="user",
    how="left"
)
```

```python
# Calculate Total Disconnect Events
total_disconnects = (
    device[device["activity"] == "Disconnect"]
    .groupby("user")
    .size()
    .reset_index(name="total_disconnects")
)
```

```python
# Merge Total Disconnect Events
device_profile = device_profile.merge(
    total_disconnects,
    on="user",
    how="left"
)
device_profile.head()
```

```python
# Calculate After-Hours Device Events Per User
after_hours_device = (
    device[device["after_hours"]]
    .groupby("user")
    .size()
    .reset_index(name="after_hours_device_events")
)
```

```python
# Merge After-Hours Device Events
device_profile = device_profile.merge(
    after_hours_device,
    on="user",
    how="left"
)
device_profile.head()
```

```python
# Calculate Weekend Device Events Per User
weekend_device = (
    device[device["weekend_activity"]]
    .groupby("user")
    .size()
    .reset_index(name="weekend_device_events")
)
weekend_device.head()
```

```python
# Merge Weekend Device Events
device_profile = device_profile.merge(
    weekend_device,
    on="user",
    how="left"
)
device_profile.head()
```

```python
# Fill Missing Values
device_profile = device_profile.fillna(0)
```

```python
# Display Employee Device Profile
device_profile.head()
```

```python
#check dataset information
device_profile.info()
```

```python
# Save Device Features
FEATURE_PATH = Path("../DATA/features")
FEATURE_PATH.mkdir(parents=True, exist_ok=True)

device_profile.to_csv(
    FEATURE_PATH / "device_features.csv",
    index=False
)

print("device_features.csv saved successfully!")
```
