```python
import csv
from pathlib import Path

ANSWER_PATH = Path("../DATA/raw/answers")

with open(ANSWER_PATH / "r6.2-1.csv", "r", encoding="utf-8", errors="ignore") as f:
    sample = f.read(2048)

dialect = csv.Sniffer().sniff(sample)
print("Detected delimiter:", repr(dialect.delimiter))
```
