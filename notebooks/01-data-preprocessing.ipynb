```python
import pandas as pd

logon_path = "/kaggle/input/datasets/andrihjonior/cert-insider-threat-dataset-r4-2/r4.2/logon.csv"

logon_df = pd.read_csv(logon_path)

print("Shape:", logon_df.shape)
logon_df.head()
```

```python
logon_df['date'] = pd.to_datetime(logon_df['date'])
```

```python
logon_df.info()
```

```python
# Extract time-based features
logon_df['hour'] = logon_df['date'].dt.hour
logon_df['day'] = logon_df['date'].dt.day
logon_df['month'] = logon_df['date'].dt.month
logon_df['weekday'] = logon_df['date'].dt.day_name()

# Office hours: 9 AM to 5 PM
logon_df['after_hours'] = ((logon_df['hour'] < 9) | (logon_df['hour'] > 17)).astype(int)

logon_df.head()
```
