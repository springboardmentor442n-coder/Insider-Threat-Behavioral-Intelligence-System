```python
base = '/kaggle/input/datasets/revsyko/insider-threat-cert-4-2/r4.2/'

logon = pd.read_csv(base + 'logon.csv')
device = pd.read_csv(base + 'device.csv')
email = pd.read_csv(base + 'email.csv')
http = pd.read_csv(base + 'http.csv')
file = pd.read_csv(base + 'file.csv')
psychometric = pd.read_csv(base + 'psychometric.csv')

print(logon.shape)
logon.head()
```

```python
print("Shape:", logon.shape)
print("\nColumns:", list(logon.columns))
print("\nData types:\n", logon.dtypes)
print("\nMissing values:\n", logon.isnull().sum())
print("\nDuplicate rows:", logon.duplicated().sum())
print("\nUnique users:", logon['user'].nunique())
print("Unique PCs:", logon['pc'].nunique())
print("\nActivity value counts:\n", logon['activity'].value_counts())
# unique() = shows you the list of values with repetitions removed
# nunique() = number of unique values → just counts how many distinct values are in that list
```

```python
print("Shape:", device.shape)
print("\nColumns:", list(device.columns))
print("\nData types:\n", device.dtypes)
print("\nMissing values:\n", device.isnull().sum())
print("\nDuplicate rows:", device.duplicated().sum())
print("\nUnique users:", device['user'].nunique())
print("Unique PCs:", device['pc'].nunique())
print("\nActivity value counts:\n", device['activity'].value_counts())
```

```python
print("Shape:", email.shape)
print("\nColumns:", list(email.columns))
print("\nData types:\n", email.dtypes)
print("\nMissing values:\n", email.isnull().sum())
print("\nDuplicate rows:", email.duplicated().sum())
print("\nUnique users:", email['user'].nunique())
if 'to' in email.columns:
    print("\nSample 'to' values:\n", email['to'].head())
if 'activity' in email.columns:
    print("\nActivity value counts:\n", email['activity'].value_counts())
```

```python
print("Shape:", http.shape)
print("\nColumns:", list(http.columns))
print("\nData types:\n", http.dtypes)
print("\nMissing values:\n", http.isnull().sum())
print("\nDuplicate rows:", http.duplicated().sum())
print("\nUnique users:", http['user'].nunique())
if 'url' in http.columns:
    print("\nSample URLs:\n", http['url'].head())
if 'activity' in http.columns:
    print("\nActivity value counts:\n", http['activity'].value_counts())
```

```python
print("Shape:", file.shape)
print("\nColumns:", list(file.columns))
print("\nData types:\n", file.dtypes)
print("\nMissing values:\n", file.isnull().sum())
print("\nDuplicate rows:", file.duplicated().sum())
print("\nUnique users:", file['user'].nunique())
if 'filename' in file.columns:
    print("\nSample filenames:\n", file['filename'].head())
if 'activity' in file.columns:
    print("\nActivity value counts:\n", file['activity'].value_counts())
```

```python
print("Shape:", psychometric.shape)
print("\nColumns:", list(psychometric.columns))
print("\nData types:\n", psychometric.dtypes)
print("\nMissing values:\n", psychometric.isnull().sum())
print("\nDuplicate rows:", psychometric.duplicated().sum())
print("\nUnique users:", psychometric['user_id'].nunique() if 'user_id' in psychometric.columns else psychometric.iloc[:,0].nunique())
print("\nSummary statistics:\n", psychometric.describe())
psychometric.head()
```
